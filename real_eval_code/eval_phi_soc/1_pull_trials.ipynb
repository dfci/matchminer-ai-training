```python
import pandas as pd
import numpy as np
trials = pd.read_csv('../../v20_public_data/trial_space_lineitems.csv').reset_index(drop=True)


trials.info()
```

```python
training_trials = pd.read_csv('../../v20_public_data/sample_trial_space_lineitems.csv')
training_trials.info()
```

```python
remaining_trials = trials[~trials.nct_id.isin(training_trials.nct_id)]
remaining_trials.info()
```

```python
unique_trial_sample = remaining_trials.groupby('nct_id').first().reset_index()[['nct_id']].sample(n=500, random_state=42)
```

```python
sample_spaces = remaining_trials[remaining_trials.nct_id.isin(unique_trial_sample.nct_id)]
```

```python
sample_spaces.info()
```

```python
sample_spaces.to_csv('sample_spaces.csv')
```
